```python
#BAI6
import math
def tam_giac(a,b,c):
    s= max(a,b,c)
    if a + b + c - s > s:
        cos = (a**2 + b**2 - c**2)/(2*a*b)
        dien_tich = 0.5*a*b*math.sqrt(1-cos**2)
        print(f"{dien_tich:.1f}")
    else:
        print("Khong phai 3 canh cua tam giac")
    
a,b,c = map(float,input("Nhap 3 canh cua tam giac: ").split())
tam_giac(a,b,c)
```
